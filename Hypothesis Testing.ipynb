```python

import math
from scipy.stats import norm
```

Given Data

```python
sample_mean = 3050
production_mean= 600
production_std= 25
sample_size= 25
cost_per_unit=5
fixed_cost=1000
alpha=0.05
```

Calculate theoretical mean weekly cost


```python
theoretical_mean = fixed_cost+cost_per_unit*production_mean
print("Theoretical mean cost:", theoretical_mean)
```

Calculate standard deviation of weekly costs

```python
sigma= cost_per_unit*production_mean
```

Calculate standard error

```python
standard_error= sigma/(sample_size**0.5)
print("Standard error:", standard_error)
```

State hypotheses

```python
print("Null hypothesis: Mean cost= Rs.4000")
print("Alternative hypothesis: Mean cost> Rs.4000")
```

Calculate Z-test statistic

```python
z_statistic =(sample_mean- theoretical_mean)/standard_error
print("Z statistic:" ,z_statistic)
```

Calculate critical Z-value

```python
critical_value = norm.ppf(1-alpha)
print("Critical Z-value:", critical_value)
```

Calculate p-value for the right tailed test

```python
p_value= norm.sf(z_statistic)
print("P-value:", p_value)
```

Make decision

```python
if z_statistic > critical_value:
    print("Reject the null Hypothesis.")
else:
    print("Fail to reject the null Hypothesis.")
```

```python
if p_value < alpha:
    print("Conclusion:Evidence supports higher weekly costs.")
else:
    print("Conclusion: Insufficient evidence that weekly costs")
    print("are higher than the theoretical model predicts")
```

Hypothesis test conducted at a 5% significance level, we fail to reject the null hypothesis. The calculated Z-statistic is -38, which is less than the critical Z-value of 1.645. The p-value is approximately 1.0, which is greater the significance level 0f 0.05.
There is insufficient statistical evidence to conclude that the average weekly operating cost of Bombay Hospitality Ltd. franchises is higher than the cost predicted by the theoretical model.

The objective of this analysis was to determine whether the average weekly operating costs of Bombay Hopitality Ltd. franchise are higher than the cost predictedd by the theoretical cost model.
